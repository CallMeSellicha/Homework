```python
# linear
def linear_search(arr, value):
    for i in range(len(arr)):
        if arr[i] == value:
            return i
    return -1

#binary
def binary_searcH(arr, value):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == value:
            return mid
        elif arr[mid] < value:
            left = mid + 1
        else:
            right = mid - 1
    return -1

# Bubble
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
    for j in range(0, n-i-1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j] #Swap
            swapped = True
    if not swapped:
        break

#selection
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
    for j in range(i+1, n):
        if arr[j] < arr[min_idx]:
            min_idx = j # Update min_idx
    arr[i], arr[min_idx] = arr[min_idx], arr[i]
```

W9A1 Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ thuật toán sắp xếp nào 

```python
def merge_sorted(a, b):
    i, j = 0, 0
    result = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    while i < len(a):
        result.append(a[i])
        i += 1
    while j < len(b):
        result.append(b[j])
        j += 1
    return result
```

W9A2
Cho ma trận hai chiều nxm có các phần tử được sắp xếp trên dòng tăng dần và trên cột
tăng dần, hãy tìm số lớn thứ k (k < nxm) với thuật toán tốt nhất có thể

```python
def find_kth_largest(matrix, k):
    n = len(matrix)
    m = len(matrix[0])
    # Đổi sang bài toán tìm số nhỏ thứ x
    target_index = n * m - k + 1 
    
    def count_less_equal(mid):
        count = 0
        row = n - 1
        col = 0
        while row >= 0 and col < m:
            if matrix[row][col] <= mid:
                count += (row + 1) # Tất cả các ô phía trên đều nhỏ hơn hoặc bằng mid
                col += 1
            else:
                row -= 1
        return count

    low = matrix[0][0]
    high = matrix[n-1][m-1]
    
    while low < high:
        mid = (low + high) // 2
        if count_less_equal(mid) < target_index:
            low = mid + 1
        else:
            high = mid
            
    return low
```

nguyên nhỏ nhất có thể 


```python
def swap(n):
    s = list(str(n))
    length = len(s)

    for i in range(length):
        minn = i
        for j in range(i + 1, length):
            if s[j] <= s[minn]:
                minn = j

        # nếu tìm được số nhỏ hơn thì swap
        if s[minn] < s[i]:
            s[i], s[minn] = s[minn], s[i]
            break
    return int("".join(s))
```

(có phải là dãy con liên tiếp của list2 hay không)? 


```python
def is_sublist(list1, list2):
    n = len(list1)
    m = len(list2)

    # nếu list1 dài hơn thì chắc chắn sai
    if n > m:
        return False

    for i in range(m - n + 1):
        match = True

        for j in range(n):
            if list2[i + j] != list1[j]:
                match = False
                break

        if match:
            return True

    return False
```

W9A8: [Phát hiện đường đi lặp lại của Rô-bốt]
Ban đầu, một con rô-bốt ở vị trí (0, 0) trên một lưới 2D.
Cho một chuỗi `moves` chứa các thao tác di chuyển của rô-bốt.
Các bước di chuyển được biểu diễn bởi các ký tự:
- 'R' (Phải)
- 'L' (Trái)
- 'U' (Lên)
- 'D' (Xuống)
Viết hàm `def is_path_crossing(moves)`:
Hàm nhận đầu vào là một chuỗi `moves`.
Hàm trả về `True` nếu trong quá trình di chuyển, rô-bốt đi vào một
ô mà nó đã từng đi qua trước đó.
Hàm trả về `False` nếu rô-bốt không bao giờ đi lặp lại (luôn đi vào
ô mới).
Lưu ý: Vị trí ban đầu (0, 0) được tính là vị trí đầu tiên đã đi qua. 


```python
def is_path_crossing(moves):
    x, y = 0, 0
    visited = {(0, 0)}

    direction = {
        'R': (0, 1),
        'L': (0, -1),
        'U': (1, 0),
        'D': (-1, 0)
    }
    
    for move in moves:
        dx, dy = direction[move]
        x += dx
        y += dy
        
        if (x, y) in visited:
            return True
        
        visited.add((x, y))
        
    return False

moves = str(input())
print(is_path_crossing(moves))
```

Viết hàm `def Median(nums):`
Hàm nhận đầu vào là một danh sách (list) `nums` được đảm bảo có đúng 5 số nguyên.
Yêu cầu:
Tìm và trả về số trung vị (số lớn thứ ba) của 5 số này.
Ràng buộc:
1. Chỉ được sử dụng tối đa 6 phép so sánh (ví dụ: <, >, <=, >=).
2. KHÔNG ĐƯỢC DÙNG bất kỳ hàm sắp xếp nào (như `sorted()`, `list.sort()`).
Chú ý: Bạn chỉ cần viết hàm `Median` theo yêu cầu.
Phần đọc input và gọi hàm (hàm main) đã có sẵn. 

```python
def Median(nums):
    a, b, c, d, e = nums

    if a > b: a, b = b, a      # 1
    if c > d: c, d = d, c      # 2
    if a > c: a, c = c, a      # 3
    if b > d: b, d = d, b      # 4
    if b > c: b, c = c, b      # 5
    if e < b:                  # 6
        return b
    else:
        return c
```

W9A11 & A12: Viết hàm factorial(n) tính giai thừa của n bằng đệ quy. Viết hàm sum_to_n(n) tính tổng từ 1 đến n bằng đệ quy.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

def sum_to_n(n):
    if n == 0:
        return 0
    return n + sum_to_n(n - 1)
```
