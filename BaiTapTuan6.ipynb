W6A2 Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị
trí i), giữ nguyên thứ tự


```python
def tong_tich_luy(a):
    kq = []
    tong = 0
    for x in a:
        tong += x
        kq.append(tong)
    return kq
a = list(map(int, input().split()))
print(tong_tich_luy(a))
```

W6A3 Nhập một tuple số nguyên (một dòng) và số nguyên k (dòng kế). In ra tuple
sau khi xoay trái k vị trí. (Giữ nguyên thứ tự tương đối sau xoay; không sắp
xếp.)


```python
def xoay_trai(t, k):
    n = len(t)
    if n == 0:
        return t
    k = k % n
    return t[k:] + t[:k]

t = tuple(map(int, input().split()))
k = int(input())

print(xoay_trai(t, k))
```

W6A4 Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp
lại, thêm vào list của key đó 

```python
def tao_dict(s):
    d = {}
    for item in s.split():
        key, value = item.split(':')
        if key not in d:
            d[key] = []
        d[key].append(value)
    return d

s = input()
print(tao_dict(s))
```

W6A5 Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros':
...} (đếm số >0, <0, =0)


```python
def dem_so(a):
    kq = {'positives': 0, 'negatives': 0, 'zeros': 0}
    
    for x in a:
        if x > 0:
            kq['positives'] += 1
        elif x < 0:
            kq['negatives'] += 1
        else:
            kq['zeros'] += 1
    
    return kq

a = list(map(int, input().split()))
print(dem_so(a))
```

Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới
trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có
trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của
key. 

```python
def gop_dict(d1, d2):
    kq = {}
    for k in d1:
        kq[k] = d1[k]

    for k in d2:
        if k in kq:
            kq[k] += d2[k]
        else:
            kq[k] = d2[k]
    kq = dict(sorted(kq.items()))
    return kq

d1 = eval(input())
d2 = eval(input())
print(gop_dict(d1, d2))
```

W6A10 Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

```python
def find(a, k):
    n = len(a)
    kq = []
    tmp = set()

    for i in range(n):
        for j in range(i + 1, n):
            if a[i] + a[j] == k:
                cap = (a[i], a[j])
                if cap not in tmp:
                    tmp.add(cap)
                    kq.append(cap)
    kq.sort()
    return kq

a = list(map(int, input().split()))
k = int(input())
print(find(a, k))
```

W6A11
[Tách số chẵn và lẻ từ tuple]
Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa
tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên
thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt.

```python
def split(a):
    chan = []
    le = []
    for i in range(len(a)):
        if a[i] % 2 == 0:
            chan.append(a[i])
        else:
            le.append(a[i])
    return tuple(chan), tuple(le)

a = tuple(map(int, input().split()))

chan, le = split(a)
print(chan)
print(le)
```

W6A12
[Tìm phần tử xuất hiện nhiều nhất]
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list.
Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất
trong số đó.

```python
def find_max(a):
    dem = {}
    for x in a:
        if x not in dem:
            dem[x] = 1
        else:
            dem[x] += 1
    maxx = max(dem.values())
    kq = min(x for x in dem if dem[x] == maxx)
    return kq

a = list(map(int, input().split()))
print(find_max(a))
```

W6A14 [Giao của hai danh sách]
Nhập vào hai dòng, mỗi dòng là một danh sách các số nguyên. Hãy tìm các
phần tử chung (giao) của hai danh sách này và in ra một list mới chứa các
phần tử chung đó, không trùng lặp. 


```python
def giao(a,b):
    kq = []
    b_set = set(b)
    for x in a:
        if x in b_set and x not in kq:
            kq.append(x)
    return kq
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(giao(a,b))
```

W6A15
[Lọc dictionary theo giá trị]
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp keyvalue từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới. 

```python
def loc_dict(a, k):
    d = {}
    for i in range(0, len(a), 2):
        key = a[i]
        value = int(a[i + 1])
        if value > k:
            d[key] = value
    return d

a = input().split()
k = int(input())

print(loc_dict(a, k))
```
