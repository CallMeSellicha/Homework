Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ
nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.
Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]


```python
a = list(map(int, input().split()))
b = []
s = set()
for i in a:
    if i not in s:
        b.append(i)
        s.add(i)
print(b)
```
