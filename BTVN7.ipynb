Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
def mangdep(a):
    dem = {}
    for x in a:
        dem[x] = dem.get(x, 0) + 1 
    loaibo = 0
    for x in dem:
        if dem[x] < x:
            loaibo += dem[x]
        else:
            loaibo += dem[x] - x
    return loaibo
n  = list(map(int,input().split()))
ans = mangdep(n)
if (ans == 0):
    print("Là mảng đẹp")
else:
    print("Loại bỏ", ans, "để thành mảng đẹp")
```

W7A17
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
def boyer_moore(a):
    pt = None
    dem = 0
    
    for x in a:
        if dem == 0:
            pt = x
        if x == pt:
            dem += 1
        else:
            dem -= 1
    if pt is not None and a.count(pt) > len(a) // 2:
        return pt
    return None
n  = list(map(int,input().split()))
ans = boyer_moore(n)
if ans is not None:
    print(ans)
```

W7A18
Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng

```python
def sapxep(a):
    chan = []
    le = []
    
    for x in a:
        if x % 2 == 0:
            chan.append(x)
        else:
            le.append(x)

    chan.sort()        
    le.sort(reverse=True)
    return chan + le

a = list(map(int,input().split()))
print(sapxep(a))
```

W7A19
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào

```python
def merge2(a, b):
    tang = True
    if len(a) >= 2 and a[0] > a[1]:
        tang = False
    
    i, j = 0, 0
    ans = []
    
    while i < len(a) and j < len(b):
        if tang: 
            if a[i] <= b[j]:
                ans.append(a[i])
                i += 1
            else:
                ans.append(b[j])
                j += 1
        else:  
            if a[i] >= b[j]:
                ans.append(a[i])
                i += 1
            else:
                ans.append(b[j])
                j += 1
    
    while i < len(a):
        ans.append(a[i])
        i += 1

    while j < len(b):
        ans.append(b[j])
        j += 1
    
    return ans

a = list(map(int,input().split()))
b = list(map(int,input().split()))
print(merge2(a, b))
```

Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu
của năm đó nếu đó là ngày hợp lệ


```python
def tim(n,t):
    songay = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    
    if t < 1 or t > 12:
        return "Ngày không hợp lệ"
    
    if n < 1 or n > songay[t - 1]:
        return "Ngày không hợp lệ"
    
    tong = 0
    for i in range(t - 1):
        tong += songay[i]
    tong += n
    return tong

n = int(input())
t = int(input())
print(tim(n,t))
```

W7A22
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
def ten_xuat_hien_nhieu_nhat(ds_ten):
    dem = {}
    
    for ho_ten in ds_ten:
        ten = ho_ten.strip().split()[-1]
        if ten in dem:
            dem[ten] += 1
        else:
            dem[ten] = 1
    
    max_lan = max(dem.values())
    
    ds_max = [ten for ten in dem if dem[ten] == max_lan]
    
    return min(ds_max)

n = int(input())
ds = [input() for _ in range(n)]

print(ten_xuat_hien_nhieu_nhat(ds))
```

W7A23
Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài >=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí
tự cái

```python
def check_ten(x):
    if len(x) < 2:
        return False
    
    if not x.isalpha():
        return False
    
    if not (x[0].isupper() and x[1:].islower()):
        return False
    
    return True

def count(s):
    ds_tu = s.split()
    dem = 0
    for x in ds_tu:
        if check_ten(x):
            dem += 1
    
    return dem

s = input()
print(count(s))
```

W7A24 Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên


```python
def delete(s):
    ds = s.split()
    if not ds:
        return ""
    
    kq = [ds[0]]  # giữ từ đầu tiên
    
    for i in range(1, len(ds)):
        if ds[i] != ds[i - 1]:
            kq.append(ds[i])
    
    return " ".join(kq)

s = input()
print(delete(s))
```

W7A25
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
def max_k(s, k):
    stack = []
    remove = len(s) - k

    for x in s:
        while remove > 0 and stack and stack[-1] < x:
            stack.pop()
            remove -= 1
        stack.append(x)

    stack = stack[:k]
    return ''.join(stack)

s = input().strip()
k = int(input())
print(max_k(s, k))
```

W7A26
Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các số đó

```python
def tinhtoan(s):
    nums = []
    tong = 0
    tb = 0
    temp = ""
    for x in s:
        if x.isdigit():
            temp += x
        else:
            if temp != "":
                nums.append(int(temp))
                temp = ""

    if temp != "":
        nums.append(int(temp))

    if len(nums) > 0:
        for i in nums:
            tong += i
        tb = tong / len(nums)
    return tong, tb

s = input()
print(tinhtoan(s))
```

Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không


```python
def check(s):
    stack = []    
    couple = {
        ')': '(',
        '}': '{',
        ']': '['
    }
    
    for x in s:
        if x in '({[':
            stack.append(x)
        elif x in ')}]':
            if not stack:
                return False
            if stack[-1] != couple[x]:
                return False
            stack.pop()
    
    return len(stack) == 0

s = input()
if check(s):
    print("YES")
else:
    print("NO")
```

Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
def snt(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def hopso(n):
    return n > 1 and not snt(n)

def max_hop_so(a, N):
    max_len = 0
    temp = 0
    start = 0       
    temp_start = 0 
    
    for i in range(len(a)):
        if hopso(a[i]) and a[i] > N:
            if temp == 0:
                temp_start = i
            temp += 1
        else:
            temp = 0
        
        if temp > max_len:
            max_len = temp
            start = temp_start
    return a[start:start + max_len]

n = list(map(int, input().split()))
N = int(input())

kq = max_hop_so(n, N)
print(len(kq))
print(*kq)
```

Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
def day_tanglt(a):
    n = len(a)
    max_len = 1  
    temp_len = 1    
    start = 0        
    temp_start = 0 
    
    for i in range(1, n):
        if a[i] > a[i - 1]:
            temp_len += 1
        else:
            temp_len = 1
            temp_start = i
        
        # Cập nhật dãy tốt nhất
        if temp_len > max_len:
            max_len = temp_len
            start = temp_start
    
    return a[start:start + max_len]

n = list(map(int, input().split()))
ans = day_tanglt(n)
print(len(ans))
print(*ans)
```


Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 

```python
def lis(a):
    n = len(a)
    dd = [1] * n
    pre = [-1] * n

    for i in range(n):
        for j in range(i):
            if a[j] < a[i] and dd[j] + 1 > dd[i]:
                dd[i] = dd[j] + 1
                pre[i] = j
    
    maxx = max(dd)
    vt = dd.index(maxx)
    
    day = []
    while vt != -1:
        day.append(a[vt])
        vt = pre[vt]
    
    day.reverse()
    return maxx, day

n = list(map(int, input().split()))
do_dai, day_con = lis(n)
print(do_dai)
print(day_con)
```
